```python
import matplotlib.pyplot as plt
import seaborn as sns

df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Order_Month"] = df["Order_Date"].dt.to_period("M")

monthly_revenue = df.groupby("Order_Month")["Net_Revenue"].sum()

plt.figure(figsize=(12,5))
plt.plot(monthly_revenue.index.astype(str), monthly_revenue.values, marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_revenue = df.groupby("Product_Category")["Net_Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
plt.bar(category_revenue.index, category_revenue.values)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
region_revenue = df.groupby("Region")["Net_Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
plt.bar(region_revenue.index, region_revenue.values)

plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()
```

```python
segment_revenue = df.groupby("Customer_Segment")["Net_Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
plt.bar(segment_revenue.index, segment_revenue.values)

plt.title("Revenue by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Revenue")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
top_products = df.groupby("Product")["Net_Revenue"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.barh(top_products.index, top_products.values)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
customer_orders = df["Customer_ID"].value_counts()

one_time = (customer_orders == 1).sum()
repeat = (customer_orders > 1).sum()

customer_type = ["One-Time Customers", "Repeat Customers"]
customer_count = [one_time, repeat]

plt.figure(figsize=(7,5))
plt.bar(customer_type, customer_count)

plt.title("Repeat vs One-Time Customers")
plt.xlabel("Customer Type")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
channel_revenue = df.groupby("Sales_Channel")["Net_Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
plt.bar(channel_revenue.index, channel_revenue.values)

plt.title("Revenue by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Revenue")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
status_counts = df["Order_Status"].value_counts()

plt.figure(figsize=(7,5))
plt.bar(status_counts.index, status_counts.values)

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

# Business Insights

## 1. Revenue Performance

The business generated approximately ₹18.55 crore in net revenue from 5,000 orders. Revenue performance varies across products, categories, regions, customer segments, and sales channels, helping identify the major contributors to overall business performance.

## 2. Customer Behavior

The analysis identified 858 unique customers, with approximately 81.47% being repeat customers. This indicates that repeat purchasing contributes significantly to the customer base and provides an opportunity to further strengthen customer retention.

## 3. Product Performance

Different products contribute differently to overall revenue. The top-performing products can be prioritized for inventory planning, marketing activities, and revenue-growth strategies.

## 4. Regional Performance

Revenue varies across regions, with the South region generating the highest revenue in this dataset. Comparing regional performance can help the business understand where sales are strongest and where additional attention may be required.

## 5. Customer Segment Performance

Revenue differs across Consumer, Corporate, and Small Business customers. Understanding these differences can help the business focus its products and marketing strategies on high-value customer segments.

## 6. Sales Channel Performance

Online, Retail Store, and Marketplace channels contribute differently to revenue. Comparing these channels helps the business understand which channels are generating stronger sales and where opportunities for improvement exist.

## 7. Business Opportunities

The analysis provides opportunities to improve customer retention, increase repeat purchases, focus on high-performing products and regions, and strengthen sales channels that generate higher revenue.

# Business Recommendations

1. Focus on high-performing products and categories to maintain revenue growth.

2. Encourage repeat purchases through loyalty programs and personalized offers.

3. Target one-time customers with strategies to convert them into repeat customers.

4. Improve performance in lower-performing regions and sales channels.

5. Monitor returned and cancelled orders to identify operational or customer issues.

6. Identify and retain high-value customers through targeted engagement strategies.
